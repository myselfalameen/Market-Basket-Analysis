### Introduction
Market Basket Analysis (MBA) is a powerful data mining technique used by retailers to increase sales by understanding customer purchasing patterns. It works by looking for combinations of items that occur together frequently in transactions.

In this notebook, we use the **Apriori Algorithm** to:
1. Identify the most frequently purchased items.
2. Discover association rules (e.g., 'If a customer buys A, they are likely to buy B').
3. Provide actionable insights for store layout and marketing strategies.

## Exploratory Data Analysis (EDA)

In this section, we will prepare the data for the Apriori algorithm and visualize the most frequent items. This includes grouping transactions, one-hot encoding, and plotting the top items using a treemap.

```python
# Group transactions by Member_number and Date to create a list of items per basket
transactions = [a[1]['itemDescription'].tolist() for a in list(basket.groupby(['Member_number','Date']))]

print(f"Total number of unique transactions: {len(transactions)}")
print("\nFirst 5 transactions:")
display(transactions[:5])
```

```python
# Use TransactionEncoder to convert the list of transactions into a one-hot encoded DataFrame
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)

# Convert the encoded array into a pandas DataFrame
transactions_df = pd.DataFrame(te_ary, columns=te.columns_)

print("One-hot encoded transactions DataFrame head:")
display(transactions_df.head())
```

```python
# Calculate the frequency of each item
item_counts = transactions_df.sum().sort_values(ascending=False)

# Create a DataFrame for the top 50 frequent items
item = pd.DataFrame({
    'Item': item_counts.index[:50],
    'Count': item_counts.values[:50]
})

print("Top 50 Frequent Items:")
display(item)
```

```python
# Visualize the top 50 frequent items using a treemap
fig, ax = plt.subplots(1, 1, figsize=(20, 12)) # Adjust figure size

# Set color scheme
cmap = matplotlib.cm.coolwarm

# Get upper and lower bounds for the color mapping
mini = min(item["Count"])
maxi = max(item["Count"])

# Set out color mapping limits
norm = matplotlib.colors.Normalize(vmin=mini, vmax=maxi)

# Obtain our raw colors
colors = [cmap(norm(value)) for value in item["Count"]]

# Create the TreeMap plot with Squarify
squarify.plot(sizes=item["Count"], label=item["Item"], alpha=0.8, color=colors, ax=ax)
plt.axis('off')
plt.title("Top 50 Frequent Basket Items", fontsize=24) # Adjust font size
plt.show()
```

## Machine Learning: Apriori Algorithm and Association Rules

Here, we apply the Apriori algorithm to find frequent itemsets and then generate association rules. We will set a minimum support of 0.1% to identify itemsets that appear at least that often, and a minimum lift of 1 to find rules that are more likely to occur together than by chance.

```python
# Apply the Apriori algorithm to find frequent itemsets
frequent_itemsets = apriori(transactions_df, min_support=0.001, use_colnames=True, max_len=5)
frequent_itemsets['length'] = frequent_itemsets['itemsets'].apply(lambda x: len(x))

print("Frequent Itemsets (first 50 rows):")
display(frequent_itemsets.head(50))
```
